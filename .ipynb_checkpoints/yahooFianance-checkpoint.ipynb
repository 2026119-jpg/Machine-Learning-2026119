```python
import yfinance as yf

# Define the ticker symbol for S&P 500 (^GSPC)
# For example, https://finance.yahoo.com/quote/%5EGSPC/history/?period1=1569174385&period2=1727027123
ticker_symbol = "^GSPC"

# Download historical data (adjust the period as needed, e.g., '1mo', '1y', 'max')
data = yf.download(ticker_symbol, period="max")  # 'max' for maximum data, adjust as needed

# Save the data to a CSV file
data.to_csv('SP500_historical_data.csv')

print("Data downloaded and saved as SP500_historical_data.csv")
```

You can adjust the period parameter to get different time ranges:
* '1d': 1 day
* '5d': 5 days
* '1mo': 1 month
* '1y': 1 year
* 'max': maximum available data.

```python
# For Dow Jones Industrial Average (^DJI)

import yfinance as yf

# Define the ticker symbol for Dow Jones Industrial Average (^DJI)

ticker_symbol = "^DJI"

# Download historical data (adjust the period as needed, e.g., '1mo', '1y', 'max')
data = yf.download(ticker_symbol, period="1y")  # 'max' for maximum data, adjust as needed

# Save the data to a CSV file
data.to_csv('DJI_1y_data.csv')

print("Data downloaded and saved as DJI_1y_data.csv")
```
