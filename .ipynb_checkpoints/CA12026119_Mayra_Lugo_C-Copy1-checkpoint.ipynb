```python
#Import all the libraries required to develop the project 
import numpy as np
import pandas as pd
import seaborn as sns
import re
import unicodedata
import matplotlib.pyplot as plt
%matplotlib inline
from sklearn.decomposition import PCA
sns.set(color_codes=True)
sns.set(style="whitegrid")
import yfinance as yf
```

```python
#Import warnings and ommit it 
import warnings
warnings.filterwarnings('ignore')
```

```python

# Define the ticker symbol for Nasdaq (^IXIC)

ticker_symbol = "QQQ"

# Download historical data (adjust the period as needed, e.g., '1mo', '1y', 'max')
data = yf.download("QQQ", period='max', auto_adjust=False) 

#clean the MultiIndex
if isinstance (data.columns, pd.MultiIndex):
    data.columns = data.columns.get_level_values(0)

#filtter last 10 year
data.index = pd.to_datetime(data.index)
df_10y = data.loc['2016-01-01':].copy()
# Deletting the rows with no value in close and adj close
df_10y = df_10y.dropna(subset=['Close', 'Adj Close'])

# Save the data to a CSV file
df_10y.to_csv('QQQ_10years.csv')

print(f"Dataset 10 years")
```
